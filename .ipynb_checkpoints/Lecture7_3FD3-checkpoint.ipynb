### For Loops
#### Another convenient tool often used when we want to carry out a task repeatedly.
#### Syntax: Starts with "for" (hence the name "for loop"). Then we create an artificial variable, often "i". This is the variable that stores the current value of the item in the list. Then we have the keyword "in", which is used before the object we want to loop over. Finally, we enter the object that we want to loop over. In the examples below, this object is one of the lists we have developed above. 

```python
j = 0
for i in dec_list:
    j = j + 1
    print("Value #", j, "is", i)
```

```python
# Note the artificial variable (we constructed "i" earlier) could really be anything (using "blahblah" now). 

j = 0
for blahblah in dec_list:
    j = j + 1
    print("Value #", j, "is", blahblah)
```

### List Comprehension
#### Just like loops help us when we have a repeated task, "list comprehension" can also be helpful in creating a list in which a repated set of instructions are passed over another "iterable item", such as a list. 
#### Syntax: Everything is written inside square brackets [ ] since it is a LIST comprehension. The first item is the "expression", or the calculation we want to do. We may create artificial variables, e.g., "i" in the example below. After that we have to explain what these artificial variables refer to. We do so by first writing "for" and then writing the artificial variables mentioned in our expression. Finally, mention the "iterable", e.g., num_list below. 

```python
# We just re-created num_list2 (above) in a single step using a list comprehension.
# Earlier, we created an empty list. Then, we filled this empty item-by-item in a loop. (a few lines of code)
# Here, we are using a list comprehension to fill-up the entire list in one go. (1 line of code)
# List comprehensions can use to make our code efficient. 

num_list3 = [i + 2 for i in num_list]
num_list3
```

```python
j = -1
for i in num_list:
    j = j+1
    k = num_list2[j]
    l = num_list3[j]
    print("num_list is", i, "num_list2 is", k, "num_list3 is", l, "i.e., same as num_list2")
```

### Dictionaries
#### Dictionaries are distinguished by curly brackets { }. A dictionary helps store data in "key-value" pairs. "Key" is like a word in a real-world dictionary and "value" is like the definition of that word. Here we have used "key" to store our column names and "value" contains all the data.
#### Syntax: Open curly brackets "{" and enter the "key", e.g., 'Period'. Add a colon ":" in front of it to define the "value" this particular key will store, e.g., we can store the list "periods". Add a comma after this if you have more keys to enter into this dictionary. Once complete, close curly brackets "}". 
#### Let's begin to work through our example. 

```python
periods = [0, 1, 2, 3, 4, 5]
cfs     = [-600, 100, 200, 300, 400, 500]

dict_npv = {'Period': periods, 'CashFlow': cfs}
dict_npv
```

```python
# We can similarly find what each "key" holds through a similar syntax as one used for lists. 
# E.g., if we wanted to find out what the key "CashFlow" holds we could write

dict_npv['Period']
```

```python
# If we want to check what keys our dictionary contains, we can use the method "keys()".

print(dict_npv.keys())
```

### **Raise to the power in Python**: In Excel, we would write = 2 ^ 3 to ask Excel to compute "2 raised to the power of 3". In Python, we would write 2**3, i.e., use two asterisks. 

```python
# Creating a new column in the dataframe to store the PV of each CF
# I can refer to each column in the data frame by writing: dfname['columnname'].
# Raise to the power in Python: In Excel, we would write =2^3 to ask Excel to compute "2 raised to the power of 3". In Python, we would
# write 2**3, i.e., use two asterisks. 

disc = 0.12

df_npv['PV'] = df_npv['CashFlow'] / ((1 + disc)**df_npv['Period'])
df_npv
```

```python
# Compute the sum of all the entries in the "PV" column using the built-in "sum()" method.

total_npv = df_npv['PV'].sum()
print(total_npv)
```

```python
# We can round this, to say 2 decimal places, as follows:

print(f"Total NPV = {total_npv:.2f}")
```

```python
# We could have used a list comprehension to do this
# See below for more explanation of the "zip()" command. 

pv_list = [ cf/(1+disc)**t for cf, t in zip(cfs, periods) ]
pv_list
```

```python
# We could've then added this to our dataframe as a new column. Not surprisingly, it has the same values as before.

df_npv['PV_v2'] = pv_list
df_npv
```

```python
# print(f"{df_npv['PV_v2'].sum():.2f}")
print(f"Total NPV = {df_npv['PV_v2'].sum():.2f}")
```

### Zip Function
#### Creates combinations of all the items that we put inside it. 
#### We can use zip various applications, such as in list comprehensions and loops.



```python
# Recall we have already defined the lists "periods" and "cfs" so no need to re-define them now. 
# Here, we are only defining a new list "cflabel" which will carry the labels for each cash flow (CF0, CF1, etc.). 

cflabel = ['CF0', 'CF1', 'CF2', 'CF3', 'CF4', 'CF5']

for t,cf,cflab in zip(periods, cfs, cflabel):
    print(cflab, ": Period = ", t, "; cash flow value = ", cf)

# [print(cflab, ": Period = ", t, "; cash flow value = ", cf) for t, cf, cflab in zip(periods,cfs,cflabel)]
```

### Creating "Data Tables" in Python. Suppose we wanted to get NPV at various discount rates. 
#### We will first create a list of all the discount rates we need. Suppose we want to go from 8% to 40% in increments of 2%. 
#### We'll create an empty list that stores our NPV results.
#### We could use a loop and list comprehension to compute the NPV at each discount rate and storing the results in our NPV results list. 
#### We can then use a dictionary (as before) to create a data frame that has two columns: discount rate and NPV. 

```python
# Creating a list of numbers (discount rates example). The BAD WAY: manually enter it.

disc_rates_list = [0.08, 0.10, 0.12, 0.14, 0.16, 0.18, 0.20, 0.22, 0.24, 0.26, 0.28, 0.30, 0.32, 0.34, 0.36, 0.38, 0.40]
```

### Numpy Library: Handy library includes many numerical operations.
#### Often abbreviated as "np"
#### "arange" method in numpy can be used to generate numbers within a range. The first argument is the start value, the second argument is "kind of like" the ending value, the third argument is increment. The ending value is never reached, which is why we have to specify a number "slightly greater" than our desired ending value to get our set of numbers. 
#### The "tolist()" command converts the resulting set of numbers into a list. We can do all this in one line as shown below. 

```python
# Creating a list of numbers (discount rates). The BETTER way: using a function with correct inputs (numpy's arange method). 

import numpy as np

disc_rates_list2 = np.arange(0.08, 0.41, 0.02).tolist()
disc_rates_list2
```

```python
# Use loop and list comprehension to store NPV at various discount rates. 
# Note: We used list comprehension which would've produced a list of PV(CF) of each CF at each discount rate. 
# We can sum it to get the NPV at each discount rate. Finally, we can save that number into our (initially empty) npv_datatable list. 

# Note to self: explain the full logic by setting up the list comprehension in front of students. 

npv_datatable = []

for discrate in disc_rates_list2:
    # npv = sum([cf/(1+discrate)**t for cf,t in zip(cfs,periods)])
    npv = sum([cf/(1+discrate)**t for cf, t in zip(cfs, periods) ])
    npv_datatable.append(npv)
    print(f"{discrate:.2f}, {npv:.2f}")
    # print(npv_datatable)
```

```python
npv_datatable
```

```python
# We can use a dictionary to create the discount rates and NPV into a data frame for further analysis. 

df_npv_datatable = pd.DataFrame({'DiscountRate': disc_rates_list2, 'NPV': npv_datatable})
df_npv_datatable
```

### More Applications: Leasing Problem
#### Suppose a machine costs 600,000 if bought today. The machine has a life of 6 years and can be resold for 100,000 at the end of year 7. Assume straight line depreciation, borrowing rate of 12% and a tax rate of 40%. Alternatively, the machine can be leased. The terms are: 6 annual payments of 140,000, with the first payment due today. 
#### Analyze the problem from the lessee's perspective. 

```python
# Lets write down all known information, just as we do in Excel. 

machine_cost = 600000
resale_value = 100000
loan_rate = 0.12
tax_rate = 0.4
depreciation = machine_cost/6
lease_value  = 140000

print(depreciation)
```

```python
# Setup a dataframe and start entering all data. Start with time periods.

periods_lease  = [0, 1, 2, 3, 4, 5, 6, 7]

df_lease = pd.DataFrame({'Periods': periods_lease})
df_lease
```

```python
# Create a new variable "MachineCost" and just set it to 0. Then manually change specific values. 
df_lease['MachineCost'] = 0
df_lease
```

```python
# Set the 1st value of this variable to -600,000. Recall that data frame index starts from 0. 

df_lease.loc[0, 'MachineCost'] = -machine_cost
df_lease
```

```python
# Add depreciation tax shields. These will arrive starting period 1 and end in period 6. 
# Start by creating a variable that only has 0s. And then add values to it. 

df_lease['Deprec_Tax_Shield'] = 0

deprec_tax_shield = tax_rate*depreciation
# print(deprec_tax_shield)

df_lease.loc[1:6, 'Deprec_Tax_Shield'] = deprec_tax_shield

df_lease
```

```python
# Following a similar approach to insert resale value in period 7 (i.e., index 7).

df_lease['Resale'] = 0
df_lease.loc[7, 'Resale'] = resale_value
df_lease
```

```python
# Resale Tax: easier since it is directly linked to an already inserted variable, "Resale". 

df_lease['ResaleTax'] = -tax_rate*df_lease['Resale']
df_lease
```

```python
# Lease payments go from period 0 to 5 (i.e., index 0-5).

df_lease['Lease'] = 0
df_lease.loc[0:5, 'Lease'] = -lease_value
df_lease
```

```python
# Lease Tax Shield: Similarly easier, since linked to an already computed variable. 

df_lease['LeaseTaxShield'] = df_lease['Lease']*-tax_rate
df_lease
```

```python
# Finally, computing Lease - Buy CFs.

df_lease['Lease_Minus_Buy'] = (df_lease['Lease'] + df_lease['LeaseTaxShield']) - (df_lease['MachineCost'] + df_lease['Deprec_Tax_Shield'] + df_lease['Resale'] + df_lease['ResaleTax'])
df_lease
```

```python
# Computing Lease - Buy IRR. Converting it into percent by multiplying by 100

irr_lease_minus_buy = (npf.irr(df_lease['Lease_Minus_Buy']))*100
print(f"Lease - Buy IRR: {irr_lease_minus_buy:.2f}")
```

```python
# After- Tax Borrowing Rate. Also in percent.

loan_rate_posttax = (loan_rate*(1 - tax_rate))*100
print(f"After-Tax Borrowing Rate: {loan_rate_posttax:.2f}")
```

### **If Condition**: Considers a statement in front of "if" and assess if it's true. If it is true, then it runs the line(s) code below it. Otherwise, it skips the line(s) code below it altogether. 

```python
# Lease or Buy Decision?
# Automating the answer using an "if condition".
# Since, After-Tax Rate < IRR >>> The first if condition is never executed.
# Recall, we just defined the variables "loan_rate_posttax" and "irr_lease_minus_buy" which store the after-tax borrowing rate and the IRR, respectively. 

if loan_rate_posttax > irr_lease_minus_buy:
    print("Leasing, since lease offers cheaper financing to lessee than borrowing from the bank")

if loan_rate_posttax < irr_lease_minus_buy:
    print("Buy, since leasing offers more expensive financing to lessee than borrowing from the bank")
```
